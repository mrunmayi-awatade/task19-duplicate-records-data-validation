```python
# Creating a sample dataset with duplicate records
import pandas as pd
data = {
    "Name": ["Amit", "Priya", "Rahul", "Amit", "Sneha", "Karan", "Priya"],
    "Age": [21, 22, 20, 21, 23, 24, 22],
    "Marks": [85, 90, 78, 85, 92, 65, 90],
    "City": ["Pune", "Mumbai", "Pune", "Pune", "Delhi", "Delhi", "Mumbai"]
}
df = pd.DataFrame(data)
print(df)
```

```python
# Identifying duplicate rows
print(df.duplicated())
```

```python
# Counting duplicate records
duplicate_count = df.duplicated().sum()
print("Number of duplicate records:", duplicate_count)
```

```python
# Displaying duplicate rows
duplicate_rows = df[df.duplicated()]
print(duplicate_rows)
```

```python
# Removing duplicate records
cleaned_df = df.drop_duplicates()
print(cleaned_df)
```

```python
# Keeping the last occurrence of duplicate records
cleaned_df = df.drop_duplicates(keep="last")
print(cleaned_df)
```

```python
# Checking missing values in the dataset
print(df.isna().sum())
```

```python
# Checking whether ages are within the valid range
valid_age = df[df["Age"].between(18, 60)]
print(valid_age)
```

```python
# Checking whether marks are within the valid range
valid_marks = df[df["Marks"].between(0, 100)]
print(valid_marks)
```

```python
# Checking whether cities belong to the allowed list
valid_cities = df[df["City"].isin(["Pune", "Mumbai", "Delhi"])]
print(valid_cities)
```

```python
# Identifying records with invalid age or marks
invalid_records = df[(~df["Age"].between(18, 60)) | (~df["Marks"].between(0, 100))]
print(invalid_records)
```

```python
# Removing duplicates and validating the data
cleaned_df = df.drop_duplicates()
cleaned_df = cleaned_df[cleaned_df["Age"].between(18, 60)]
cleaned_df = cleaned_df[cleaned_df["Marks"].between(0, 100)]
cleaned_df = cleaned_df[cleaned_df["City"].isin(["Pune", "Mumbai", "Delhi"])]
print(cleaned_df)
```
